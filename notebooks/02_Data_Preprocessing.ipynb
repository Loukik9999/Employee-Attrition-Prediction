```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Save objects
import joblib
```

```python
df = pd.read_csv("IBM_EmpDataset.csv")
```

```python
df.drop(columns=["EmployeeNumber"], inplace=True)
```

```python
# Separate Features and Target
X = df.drop(columns=["Attrition", "Attrition_Flag"])
y = df["Attrition"]
```

```python
# Identify Numerical and Categorical Columns
numerical_features = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = X.select_dtypes(include=["object"]).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
# Encoding
encoder = OneHotEncoder(
    drop="first",
    sparse_output=False,
    handle_unknown="ignore"
)
```

```python
encoded_array = encoder.fit_transform(X[categorical_features])
encoded_array
```

```python
# Create DataFrame
encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out(categorical_features),
    index=X.index
)
```

```python
# Remove Original Categorical Columns
X = X.drop(columns=categorical_features)
```

```python
# Add Encoded Features
X = pd.concat([X, encoded_df], axis=1)
```

```python
X.head()
```

```python
# Encode Target Variable
y = df["Attrition_Flag"]
```

```python
y.value_counts()
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training Data :", X_train.shape)
print("Testing Data :", X_test.shape)
```

```python
# Standard Scaling
scaler = StandardScaler()
```

```python
X_train[numerical_features] = scaler.fit_transform(
    X_train[numerical_features]
)
```

```python
X_test[numerical_features] = scaler.transform(
    X_test[numerical_features]
)
```

```python
X_train.head()
```

```python
joblib.dump(encoder, "onehot_encoder.pkl")
joblib.dump(scaler, "standard_scaler.pkl")
joblib.dump(numerical_features, "numerical_features.pkl")
joblib.dump(categorical_features, "categorical_features.pkl")
```

```python
X_train.to_csv("X_train.csv", index=False)
X_test.to_csv("X_test.csv", index=False)

y_train.to_csv("y_train.csv", index=False)
y_test.to_csv("y_test.csv", index=False)
```
